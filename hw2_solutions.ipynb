#### 19. Can the missing values in the `time` column be inferred from other available data? If so, fill in the missing values appropriately.

```python
# Let's check if features 'datePostedString' and 'time' are consistent:

# First let's convert time string to time
df['datePostedString'] = pd.to_datetime(df['datePostedString'], errors='coerce')

print(np.sum(df['time'] != df['datePostedString']))
print(np.sum(df['time'] == df['datePostedString']))
```

They are not always consistent. Let's examine the inconsistencies.

```python
df.loc[df['time'] != df['datePostedString'], ['time','datePostedString']].sample(10)
```

The date in the `time` column can be larger, equal, or smaller than `datePostedString`. The difference could be remarkable.

```python
df.loc[(df['time'] - df['datePostedString'] == np.max(df['time'] - df['datePostedString'])), ['time','datePostedString']]
```

```python
df.loc[(df['time'] - df['datePostedString'] == np.min(df['time'] - df['datePostedString'])), ['time','datePostedString']]
```

I would say it is NOT advisable to replace missing values in the `time` column with `datePostedString`. However, if we decide to do so, we could use the following commands:

```python
df.loc[df['time'].isnull(), 'time'] = pd.to_datetime(df.loc[df['time'].isnull(), 'datePostedString'])
```

#### Can we reliably infer missing `cityId` values using `city`?

To answer such question, we need to understand a few things.

#### 25. Do all cities have unique `cityId`? Find cities wuth multiple IDs if any.

```python
unique_city_counts = df.groupby('city')['cityId'].nunique() # .nunique() counts only non-null unique values by default
unique_city_counts[unique_city_counts > 1]
```

#### 26. Does every city have at least one `cityId`? Identify any cities without a `cityId` if they exist.

```python
unique_city_counts[unique_city_counts  < 1]
```

#### 27. Replace `NaN` values in the `cityId` column with the corresponding `cityId` for cities that have a unique `cityId`. Report how many missing values where in the `cityId` column before and after the replacement.

```python
print(f'before {df['cityId'].isna().sum()}')

cities_with_one_id = unique_city_counts[unique_city_counts == 1].index
city_to_cityId = df[df['city'].isin(cities_with_one_id)].groupby('city')['cityId'].max()

df['cityId'] = df['cityId'].fillna(df['city'].map(city_to_cityId))

print(f'after {df['cityId'].isna().sum()}')
```
